```python
import pandas as pd

# Identify columns containing years
year_cols = [col for col in df.columns if "[YR" in col]

# Convert year columns to numeric values
df[year_cols] = df[year_cols].replace("..", pd.NA)

df[year_cols] = df[year_cols].apply(
    pd.to_numeric, errors="coerce"
)
```
